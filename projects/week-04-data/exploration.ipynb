```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

rng = np.random.default_rng(42)

print("NumPy version:", np.__version__)
print("环境准备完成！")
```

```python
python_list = [1, 2, 3, 4, 5]
numpy_array = np.array([1, 2, 3, 4, 5])

print("Python list:", python_list)
print("NumPy array:", numpy_array)

print("list * 2:", python_list * 2)
print("array * 2:", numpy_array * 2)
```

```python
scores = np.array([58, 72, 89, 45, 96, 67])
condition = scores >= 60
filtered_scores = scores[condition]
print(filtered_scores)
```

```python
scores = np.array([58, 72, 89, 45, 96, 67, 80, 60])
condition_part01 = (scores >= 60) & (scores <= 80)
filtered_scores = scores[condition_part01]
print(filtered_scores)
```

```python
print("总人数:", scores.size)
print("平均分:", scores.mean())
print("最高分:", scores.max())
print("最低分:", scores.min())
print("总分:", scores.sum())
print("标准差:", scores.std())

passed = scores >= 60

print("及格人数:", passed.sum())
print("及格率:", passed.mean())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["font.sans-serif"] = ["Microsoft YaHei", "SimHei"]
plt.rcParams["axes.unicode_minus"] = False

print(f"总人数: {scores.size}")
print(f"平均分: {scores.mean():.2f}")
print(f"最高分: {scores.max()}")
print(f"最低分: {scores.min()}")
print(f"标准差: {scores.std():.2f}")
print(f"及格人数: {passed.sum()}")
print(f"及格率: {passed.mean():.1%}")

plt.figure(figsize=(8, 5))

sns.histplot(
    scores,
    bins=[40, 50, 60, 70, 80, 90, 100],
    edgecolor="white"
)

plt.axvline(
    scores.mean(),
    color="red",
    linestyle="--",
    label=f"平均分：{scores.mean():.2f}"
)

plt.title("成绩分布")
plt.xlabel("分数")
plt.ylabel("学生人数")
plt.legend()
plt.show()

student_ids = np.arange(1, scores.size + 1)

colors = np.where(scores >= 60, "#4C9F70", "#D95D39")

plt.figure(figsize=(9, 5))
bars = plt.bar(student_ids, scores, color=colors)

plt.axhline(
    60,
    color="red",
    linestyle="--",
    label="及格线：60分"
)

plt.bar_label(bars, padding=3)

plt.title("学生成绩")
plt.xlabel("学生编号")
plt.ylabel("分数")
plt.xticks(student_ids)
plt.ylim(0, 105)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 3))

sns.boxplot(
    x=scores,
    color="#E9C46A"
)

plt.axvline(
    scores.mean(),
    color="red",
    linestyle="--",
    label=f"平均分：{scores.mean():.2f}"
)

plt.title("成绩箱线图")
plt.xlabel("分数")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["font.sans-serif"] = ["Microsoft YaHei", "SimHei"]
plt.rcParams["axes.unicode_minus"] = False

temperature_array = np.array([18, 21, 25, 29, 33, 35, 31, 27, 23, 19])

mean_temperature = temperature_array.mean()
median_temperature = np.median(temperature_array)

print("平均温度:", mean_temperature)
print("温度中位数", median_temperature)

above_average_condition = temperature_array >= temperature_array.mean()
above_thirty_degree = temperature_array >= 30

print("大于平均温度的数据:", temperature_array[above_average_condition])
print("大于三十度的天数:", above_thirty_degree.sum())

temperature_ids = np.arange(1, temperature_array.size + 1)

colors = np.where(temperature_array >= 30, "#D95D39","#4C719F")

plt.figure(figsize=(9, 5))
bars = plt.bar(temperature_ids, temperature_array, color=colors)

plt.axhline(
    temperature_array.mean(),
    color="red",
    linestyle="--",
    label="平均温度"
)

plt.bar_label(bars, padding=3)

plt.title("温度检测统计")
plt.xlabel("日期")
plt.ylabel("温度")
plt.xticks(temperature_ids)
plt.ylim(0, 40)
plt.legend()
plt.show()
```
